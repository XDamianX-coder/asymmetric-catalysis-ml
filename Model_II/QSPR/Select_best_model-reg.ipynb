```python
import pandas as pd

# Define the path to the Excel file
file_path = '../Data/Quality_ee_15_.xlsx'

# Define the sheets and metrics
sheets = ["MLR", "DT", "RF", "KNN", "SVM"]
metrics = ['Training data R^2 score', 'Test data R^2 score', 'Training RMSE', 'Test RMSE']
number_of_features = 10
# Initialize a dictionary to store the best models
best_models = {}

# Iterate through each sheet to find the best model
for sheet in sheets:
    # Read the data from the current sheet
    df = pd.read_excel(file_path, sheet_name=sheet)
    df_filtered = df[df['Number of features'] < number_of_features]
    # Filter the DataFrame based on the criteria
    # The model must have higher training parameters than test parameters
    filtered_df = df_filtered[(df_filtered['Training data R^2 score'] > df_filtered['Test data R^2 score']) & 
                     (df_filtered['Training RMSE'] < df_filtered['Test RMSE']) & 
                              (df_filtered['Test data R^2 score'] > 0.36)]
    
    if not filtered_df.empty:
        # Find the row with the maximum Test Accuracy
        best_model = filtered_df.loc[filtered_df['Test data R^2 score'].idxmax()]
        best_model = filtered_df.loc[filtered_df['Correlation threshold'].idxmax()]
        best_models[sheet] = best_model

# Convert the results to a DataFrame for better visualization
best_models_df = pd.DataFrame.from_dict(best_models, orient='index')

# Print the best models for each method
print("Best Models for Each Method:")
best_models_df
```
