```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error


df = pd.read_excel("../Data/ee_random_state_42.xlsx", sheet_name='Train')

observed = df['ee']
predicted = df['predicted ee']

# Calculate metrics
r2 = r2_score(observed, predicted)
mse = mean_squared_error(observed, predicted)
mae = mean_absolute_error(observed, predicted)
rmse = np.sqrt(mse)
#r = np.corrcoef(observed, predicted)[0, 1]
r = np.sqrt(r2)

# Create the plot
plt.figure(figsize=(6, 6), dpi=300)  # Publication-quality size and resolution

# Scatter plot
plt.scatter(observed, predicted, color='#1f77b4', alpha=0.6, s=50, label='Data points')

# 1:1 line
min_val, max_val = min(observed.min(), predicted.min()), max(observed.max(), predicted.max())
plt.plot([min_val, max_val], [min_val, max_val], 'k--', lw=1)#, label='1:1 line')

# Add metrics text box
metrics_text = (
    f'Train\n'
    f'$R^2$: {r2:.2f}\n'
    f'R: {r:.2f}\n'
    f'MSE: {mse:.2f}\n'
    f'RMSE: {rmse:.2f}\n'
    f'MAE: {mae:.2f}'
)
plt.text(
    0.05, 0.90, metrics_text,
    transform=plt.gca().transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='white', alpha=0.7, edgecolor='black')
)

# Customize plot
plt.xlabel('Observed Enantiomeric Excess', fontsize=12)
plt.ylabel('Predicted Enantiomeric Excess', fontsize=12)
#plt.title('Predicted vs. Observed Values', fontsize=14, pad=10)
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.7)

# Ensure equal aspect ratio
plt.gca().set_aspect('equal', adjustable='box')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Save as high-quality PDF
#plt.savefig('predicted_vs_observed_train.pdf', format='pdf', bbox_inches='tight')
plt.show()
# Close the plot to free memory
plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Load train data
df_train = pd.read_excel("../Data/ee_random_state_42.xlsx", sheet_name='Train')
observed_train = df_train['ee']
predicted_train = df_train['predicted ee']

# Load test data
df_test = pd.read_excel("../Data/ee_random_state_42.xlsx", sheet_name='Test')
observed_test = df_test['ee']
predicted_test = df_test['predicted ee']

# Calculate metrics for train
r2_train = r2_score(observed_train, predicted_train)
mse_train = mean_squared_error(observed_train, predicted_train)
mae_train = mean_absolute_error(observed_train, predicted_train)
rmse_train = np.sqrt(mse_train)
r_train = np.sqrt(r2_train)

# Calculate metrics for test
r2_test = r2_score(observed_test, predicted_test)
mse_test = mean_squared_error(observed_test, predicted_test)
mae_test = mean_absolute_error(observed_test, predicted_test)
rmse_test = np.sqrt(mse_test)
r_test = np.sqrt(r2_test)

# Create one combined plot
plt.figure(figsize=(7,7), dpi=300)

# Scatter plots with different colors
plt.scatter(observed_train, predicted_train, color='#1f77b4', alpha=0.6, s=50, label='Train')
plt.scatter(observed_test, predicted_test, color='#ff7f0e', alpha=0.6, s=50, label='Test')

# 1:1 line covering min/max of combined data
min_val = min(observed_train.min(), predicted_train.min(), observed_test.min(), predicted_test.min())
max_val = max(observed_train.max(), predicted_train.max(), observed_test.max(), predicted_test.max())
plt.plot([min_val, max_val], [min_val, max_val], 'k--', lw=1)

# Metrics text box - Train
metrics_text_train = (
    f'Train\n'
    f'$R^2$: {r2_train:.2f}\n'
    f'R: {r_train:.2f}\n'
    f'MSE: {mse_train:.2f}\n'
    f'RMSE: {rmse_train:.2f}\n'
    f'MAE: {mae_train:.2f}'
)
# Metrics text box - Test
metrics_text_test = (
    f'Test\n'
    f'$R^2$: {r2_test:.2f}\n'
    f'R: {r_test:.2f}\n'
    f'MSE: {mse_test:.2f}\n'
    f'RMSE: {rmse_test:.2f}\n'
    f'MAE: {mae_test:.2f}'
)

# Add train metrics box
plt.text(
    0.05, 0.95, metrics_text_train,
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='#1f77b4', alpha=0.2, edgecolor='none')
)

# Add test metrics box
plt.text(
    0.70, 0.05, metrics_text_test,
    transform=plt.gca().transAxes,
    fontsize=12,
    verticalalignment='bottom',
    bbox=dict(boxstyle='round', facecolor='#ff7f0e', alpha=0.2, edgecolor='none')
)

# Customize plot
plt.xlabel('Observed Enantiomeric Excess', fontsize=14)
plt.ylabel('Predicted Enantiomeric Excess', fontsize=14)
plt.legend(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.gca().set_aspect('equal', adjustable='box')

plt.tight_layout()
#plt.savefig('predicted_vs_observed_train_test_combined.pdf', format='pdf', bbox_inches='tight')
plt.savefig('Figures/predicted_vs_observed_train_test_combined.svg', format='svg', bbox_inches='tight')
plt.show()
plt.close()
```
