```python
import sys
from pathlib import Path
prediction_mode_path = Path("../module")
sys.path.append(prediction_mode_path.as_posix())
import models_creation as pred_model

import pandas as pd

from sklearn.model_selection import train_test_split

import joblib

from sklearn.metrics import r2_score, mean_squared_error
import math

import numpy as np
```

```python
smiles_df = pd.read_excel('../Data/QSPR_epoxidation_model_III.xlsx')
```

```python
smiles_df.head()
```

```python
molecular_descriptors_df = pred_model.prepare_data('../Data/QSPR_epoxidation_model_III.xlsx')
molecular_descriptors_df.head()
```

```python
frames = [smiles_df['SMILES'], molecular_descriptors_df]
res = pd.concat(frames, axis=1)
```

```python
res.head()
```

```python
## load models
model_ee = joblib.load('models/random_forest_model_8_estimators_EE.joblib') #Random state 15
```

# Construct a dataframe for each Random Seed and predict values, additionally calculate R score and RMSE...

## EE

```python
list(model_ee.feature_names_in_)
```

```python
res.columns[0:-9]
```

```python
X_train, X_test, y_train, y_test = train_test_split(res[res.columns], 
                                                        res['ee'], 
                                                        test_size=0.15, random_state=15)

X_train['predicted ee'] = model_ee.predict(X_train[list(model_ee.feature_names_in_)])
X_train['observed ee'] = y_train

X_test['predicted ee'] = model_ee.predict(X_test[list(model_ee.feature_names_in_)])
X_test['observed ee'] = y_test


print('Train R score: '+str(math.sqrt(r2_score(X_train['observed ee'], X_train['predicted ee']))))
print('Test R score: '+str(math.sqrt(r2_score(X_test['observed ee'], X_test['predicted ee']))))
print('Train RMSE score: '+str(math.sqrt(mean_squared_error(X_train['observed ee'], X_train['predicted ee']))))
print('Test RMSE score: '+str(math.sqrt(mean_squared_error(X_test['observed ee'], X_test['predicted ee']))))
```

```python
with pd.ExcelWriter("../Data/ee_random_state_15.xlsx") as writer:
    X_train.to_excel(writer, sheet_name="Train", index=True)
    X_test.to_excel(writer, sheet_name="Test", index=True)
```
